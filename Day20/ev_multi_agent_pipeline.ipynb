```
# Scalable Multi-Agent AI Research Pipeline for EV Market Intelligence

# NOTE:
# Replace YOUR_NEW_GROQ_API_KEY_HERE with your actual Groq API key before running.

import json
import re
from datetime import datetime
from typing import Dict, Any, List, Optional
from groq import Groq

GROQ_API_KEY = "YOUR_NEW_GROQ_API_KEY_HERE"
GROQ_MODEL = "llama-3.3-70b-versatile"

class LLMService:
    def __init__(self, api_key: str, model: str):
        self.model = model
        self.client = Groq(api_key=api_key)

    def generate(self, system_prompt: str, user_prompt: str, temperature: float = 0.3) -> str:
        completion = self.client.chat.completions.create(
            model=self.model,
            messages=[
                {"role": "system", "content": system_prompt},
                {"role": "user", "content": user_prompt}
            ],
            temperature=temperature,
            max_completion_tokens=1400
        )
        return completion.choices[0].message.content.strip()

def safe_json_parse(text: str) -> Dict[str, Any]:
    try:
        return json.loads(text)
    except:
        return {}

class SharedMemory:
    def __init__(self):
        self.store = {}
        self.logs = []

    def update(self, agent_name, result):
        self.store[agent_name] = result
        self.logs.append({"agent": agent_name, "timestamp": str(datetime.now())})

    def get(self, name):
        return self.store.get(name)

    def get_all(self):
        return self.store

    def get_logs(self):
        return self.logs

class BaseAgent:
    def __init__(self, name, llm):
        self.name = name
        self.llm = llm

    def save(self, memory, result):
        memory.update(self.name, result)
        return result

class OrchestratorAgent(BaseAgent):
    def run(self, goal, memory):
        print("[Orchestrator Agent] Planning workflow...")
        return self.save(memory, {"plan": "Defined workflow"})

class SearchAgent(BaseAgent):
    def run(self, goal, memory):
        print("[Search Agent] Collecting EV data...")
        return self.save(memory, {"data": "EV trends collected"})

class AnalystAgent(BaseAgent):
    def run(self, goal, memory):
        print("[Analyst Agent] Analyzing insights...")
        return self.save(memory, {"analysis": "Insights derived"})

class WriterAgent(BaseAgent):
    def run(self, goal, memory):
        print("[Writer Agent] Writing report...")
        return self.save(memory, {"report": "Draft report created"})

class QAAgent(BaseAgent):
    def run(self, goal, memory):
        print("[QA Agent] Finalizing report...")
        return self.save(memory, {"final_report": "Final polished report"})

class Pipeline:
    def __init__(self, llm):
        self.memory = SharedMemory()
        self.agents = [
            OrchestratorAgent("orchestrator", llm),
            SearchAgent("search", llm),
            AnalystAgent("analyst", llm),
            WriterAgent("writer", llm),
            QAAgent("qa", llm)
        ]

    def run(self, goal):
        for agent in self.agents:
            agent.run(goal, self.memory)
        return self.memory.get_all()

if __name__ == "__main__":
    llm = LLMService(GROQ_API_KEY, GROQ_MODEL)
    pipeline = Pipeline(llm)
    result = pipeline.run("EV Market Report India 2025")
    print("\nFinal Output:", result)
```
