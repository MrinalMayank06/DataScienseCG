```python
# 🧩 Scenario
# A university is receiving thousands of student queries daily (admissions, fees, hostel, courses).
# Manual support is slow and inconsistent.

# The university wants to build an AI-powered chatbot that:

# Answers student queries instantly

# Personalizes responses using Student ID

# Logs each interaction for audit

# 🎯 Your Task
# Design and implement an AI chatbot system using Azure services.

# ⚙️ Requirements
# You MUST:
# Use

# Azure OpenAI Service
# Azure AI Studio
# Create resource and deploy a model

# Use:
# API Key
# Endpoint
# Pass Student ID in every request
# Ensure chatbot response includes:
# Student ID
# Answer


from google.colab import userdata
from openai import OpenAI
import json
from datetime import datetime

# 🔐 Load secrets securely (NO hardcoding)
api_key = userdata.get("AZURE_OPENAPI_KEY")
endpoint = "https://mrinalopenai.openai.azure.com"
deployment_name = "gpt-4o"

# Initialize Azure OpenAI client
client = OpenAI(
    api_key=api_key,
    base_url=f"{endpoint}/openai/deployments/{deployment_name}/",
    default_query={"api-version": "2024-02-01"}
)

# 🎯 Chatbot function
def university_chatbot(data):
    student_id = data["student_id"]
    query = data["query"]

    prompt = f"""
    You are a university AI assistant.

    Always include Student ID in the response.

    Student ID: {student_id}
    Question: {query}
    """

    response = client.chat.completions.create(
        model=deployment_name,
        messages=[{"role": "user", "content": prompt}]
    )

    answer = response.choices[0].message.content

    # 📊 Logging (for audit)
    log = {
        "student_id": student_id,
        "query": query,
        "response": answer,
        "timestamp": str(datetime.now())
    }

    with open("chat_logs.json", "a") as f:
        f.write(json.dumps(log) + "\n")

    return f"[Student ID: {student_id}]\n{answer}"


# 📥 Input
request_data = {
    "student_id": "BTECH2026_102",
    "query": "What is the fee structure?"
}

# 🚀 Run
print(university_chatbot(request_data))
```
