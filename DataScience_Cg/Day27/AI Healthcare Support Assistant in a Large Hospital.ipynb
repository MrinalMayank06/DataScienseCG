SCENARIO: “AI Healthcare Support Assistant in a Large Hospital”
🏢 Background Story
A large hospital (like Apollo or Fortis) has thousands of patients and staff members.
👉 Daily Challenges Faced:
- Patients waiting hours for appointment confirmations
- Confusion about lab test results and reports
- Doctors overwhelmed with scheduling and follow-up reminders
- Nurses struggling to track medicine administration times
- Emergency cases needing instant triage
👉 Instead of manual coordination, the hospital builds an:
🤖 AI Healthcare Support Assistant
🚑 Capabilities:
- 📅 Smart Scheduling: Automatically books and reschedules patient appointments based on doctor availability.
- 🧪 Lab Report Explainer: Summarizes test results in simple language for patients.
- 💊 Medication Tracker: Sends reminders to nurses and patients about dosage timings.
- 🚨 Emergency Triage: Instantly categorizes incoming cases (critical, urgent, routine) and alerts the right medical team.
- 📧 Follow-up Automation: Sends personalized recovery instructions and reminders after discharge.
🎯 Impact:
- Reduced patient waiting time
- Doctors spend more time on treatment, less on admin work
- Nurses avoid errors in medicine administration
- Faster response in emergencies
- Improved patient satisfaction and trust

```python
!pip install groq gradio nest_asyncio

import os
import asyncio
import nest_asyncio
import gradio as gr
from groq import Groq

os.environ["GROQ_API_KEY"] = "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"

client = Groq(api_key=os.getenv("GROQ_API_KEY"))

nest_asyncio.apply()

_original_asyncio_run = asyncio.run

def compatible_asyncio_run(main, *, debug=None, loop_factory=None):
    return _original_asyncio_run(main)

asyncio.run = compatible_asyncio_run


patients = {
    "P101": {
        "name": "Rahul Sharma",
        "doctor": "Dr. Mehta",
        "appointment": {
            "date": "2026-04-02",
            "time": "10:30 AM",
            "department": "Cardiology",
            "status": "Confirmed"
        },
        "lab_report": {
            "test": "Blood Sugar and Cholesterol Panel",
            "summary": "Blood sugar is within normal range, but cholesterol is slightly elevated.",
            "status": "Needs lifestyle follow-up"
        },
        "medication": {
            "medicine": "Atorvastatin 10mg",
            "dose_time": "8:00 PM",
            "instruction": "Take after dinner"
        },
        "triage": {
            "symptoms": "Chest discomfort and fatigue",
            "severity": "Urgent",
            "action": "Alert cardiology duty team"
        },
        "follow_up": {
            "message": "Continue low-fat diet, stay active, and attend next cardiology review.",
            "next_review": "2026-04-09"
        }
    },
    "P102": {
        "name": "Priya Verma",
        "doctor": "Dr. Khanna",
        "appointment": {
            "date": "2026-04-03",
            "time": "02:00 PM",
            "department": "Neurology",
            "status": "Rescheduled"
        },
        "lab_report": {
            "test": "Vitamin D and Hemoglobin",
            "summary": "Vitamin D is low and hemoglobin is slightly below ideal range.",
            "status": "Supplement advice needed"
        },
        "medication": {
            "medicine": "Vitamin D3 Sachet",
            "dose_time": "9:00 AM",
            "instruction": "Take after breakfast once weekly"
        },
        "triage": {
            "symptoms": "Recurring headache and weakness",
            "severity": "Routine",
            "action": "Continue outpatient review"
        },
        "follow_up": {
            "message": "Follow supplement plan and repeat blood work in 4 weeks.",
            "next_review": "2026-04-24"
        }
    }
}


async def get_appointment_status(patient_id):
    await asyncio.sleep(1)

    if patient_id in patients:
        appt = patients[patient_id]["appointment"]
        return (
            f"Appointment Status:\n"
            f"- Patient: {patients[patient_id]['name']}\n"
            f"- Doctor: {patients[patient_id]['doctor']}\n"
            f"- Date: {appt['date']}\n"
            f"- Time: {appt['time']}\n"
            f"- Department: {appt['department']}\n"
            f"- Status: {appt['status']}"
        )

    return "Patient not found."


async def get_lab_report_summary(patient_id):
    await asyncio.sleep(1)

    if patient_id in patients:
        report = patients[patient_id]["lab_report"]
        return (
            f"Lab Report Summary:\n"
            f"- Patient: {patients[patient_id]['name']}\n"
            f"- Test: {report['test']}\n"
            f"- Summary: {report['summary']}\n"
            f"- Status: {report['status']}"
        )

    return "Patient not found."


async def get_medication_tracker(patient_id):
    await asyncio.sleep(1)

    if patient_id in patients:
        med = patients[patient_id]["medication"]
        return (
            f"Medication Tracker:\n"
            f"- Patient: {patients[patient_id]['name']}\n"
            f"- Medicine: {med['medicine']}\n"
            f"- Dose Time: {med['dose_time']}\n"
            f"- Instruction: {med['instruction']}"
        )

    return "Patient not found."


async def get_emergency_triage(patient_id):
    await asyncio.sleep(1)

    if patient_id in patients:
        triage = patients[patient_id]["triage"]
        return (
            f"Emergency Triage:\n"
            f"- Patient: {patients[patient_id]['name']}\n"
            f"- Symptoms: {triage['symptoms']}\n"
            f"- Severity: {triage['severity']}\n"
            f"- Action: {triage['action']}"
        )

    return "Patient not found."


async def get_follow_up_plan(patient_id):
    await asyncio.sleep(1)

    if patient_id in patients:
        follow = patients[patient_id]["follow_up"]
        return (
            f"Follow-up Plan:\n"
            f"- Patient: {patients[patient_id]['name']}\n"
            f"- Instruction: {follow['message']}\n"
            f"- Next Review: {follow['next_review']}"
        )

    return "Patient not found."


async def parallel_healthcare_support(patient_id):
    results = await asyncio.gather(
        get_appointment_status(patient_id),
        get_lab_report_summary(patient_id),
        get_medication_tracker(patient_id),
        get_emergency_triage(patient_id),
        get_follow_up_plan(patient_id),
        return_exceptions=True
    )

    appointment, lab_report, medication, triage, follow_up = results

    return {
        "appointment": appointment if not isinstance(appointment, Exception) else "Appointment data unavailable",
        "lab_report": lab_report if not isinstance(lab_report, Exception) else "Lab report data unavailable",
        "medication": medication if not isinstance(medication, Exception) else "Medication data unavailable",
        "triage": triage if not isinstance(triage, Exception) else "Triage data unavailable",
        "follow_up": follow_up if not isinstance(follow_up, Exception) else "Follow-up data unavailable"
    }


def decide_intent(user_query):
    response = client.chat.completions.create(
        model="llama-3.3-70b-versatile",
        messages=[{
            "role": "user",
            "content": f"""
You are an AI healthcare support assistant intent classifier.

Classify the user's query into exactly one of these:
- appointment
- lab_report
- medication
- triage
- follow_up
- full_report

Rules:
- appointment confirmation, scheduling, rescheduling -> appointment
- lab result, test report, explain report -> lab_report
- medicine, dosage, tracker, reminder -> medication
- emergency, urgent, severity, triage -> triage
- discharge, recovery plan, follow-up reminder -> follow_up
- full summary, complete report, full healthcare support summary -> full_report

Return exactly one label.

User Query: {user_query}
"""
        }]
    )
    return response.choices[0].message.content.strip().lower()


def analyse_healthcare_support_data(text, patient_id):
    patient_name = patients.get(patient_id, {}).get("name", "Patient")

    response = client.chat.completions.create(
        model="llama-3.3-70b-versatile",
        messages=[{
            "role": "user",
            "content": f"""
Analyze the following hospital support intelligence for {patient_name}.

Provide:
1. Key healthcare support summary
2. Important observations
3. Clinical or coordination note
4. Risk/priority insight
5. Simple patient/staff-friendly explanation

Data:
{text}
"""
        }]
    )
    return response.choices[0].message.content


def generate_healthcare_support_report(analysis, patient_id):
    patient_name = patients.get(patient_id, {}).get("name", "Patient")

    response = client.chat.completions.create(
        model="llama-3.3-70b-versatile",
        messages=[{
            "role": "user",
            "content": f"""
Create a professional hospital support summary report for {patient_name}.

Use this analysis:
{analysis}

Keep it:
- clear
- structured
- concise but informative
- suitable for doctors, nurses, and hospital support staff
"""
        }]
    )
    return response.choices[0].message.content


async def full_pipeline(patient_id, user_query):
    if patient_id not in patients:
        return "Invalid Patient ID. Please enter P101 or P102."

    data = await parallel_healthcare_support(patient_id)
    intent = decide_intent(user_query)

    if intent == "appointment":
        combined_text = data["appointment"]
    elif intent == "lab_report":
        combined_text = data["lab_report"]
    elif intent == "medication":
        combined_text = data["medication"]
    elif intent == "triage":
        combined_text = data["triage"]
    elif intent == "follow_up":
        combined_text = data["follow_up"]
    else:
        combined_text = f"""
{data['appointment']}

{data['lab_report']}

{data['medication']}

{data['triage']}

{data['follow_up']}
""".strip()

    analysis = analyse_healthcare_support_data(combined_text, patient_id)
    report = generate_healthcare_support_report(analysis, patient_id)

    final_output = f"""
==============================
AI HEALTHCARE SUPPORT ASSISTANT OUTPUT
==============================

Patient ID: {patient_id}
Detected Intent: {intent}

RAW DATA:
{combined_text}

--------------------------------
AI ANALYSIS + SUPPORT SUMMARY:
--------------------------------
{report}
"""
    return final_output.strip()


def run_normal_mode():
    print("AI Healthcare Support Assistant in a Large Hospital")
    patient_id = input("Enter Patient ID (P101 / P102): ").strip()
    user_question = input("Ask your question: ").strip()

    if not patient_id or not user_question:
        print("Please enter both Patient ID and question.")
        return

    result = asyncio.run(full_pipeline(patient_id, user_question))
    print("\nFINAL OUTPUT:\n")
    print(result)


def healthcare_support_ui(patient_id, user_query):
    patient_id = patient_id.strip()
    user_query = user_query.strip()

    if not patient_id or not user_query:
        return "Please enter both Patient ID and question."

    return asyncio.run(full_pipeline(patient_id, user_query))


run_normal_mode()


with gr.Blocks() as demo:
    gr.Markdown("# AI Healthcare Support Assistant in a Large Hospital")
    gr.Markdown("""
Supported Patient IDs:
- P101
- P102

Example questions:
- Confirm my appointment status.
- Explain my lab test report.
- Show my medicine timing.
- What is my emergency triage category?
- Give me my follow-up plan.
- Give me a full healthcare support summary.
""")

    patient_input = gr.Textbox(
        label="Enter Patient ID",
        placeholder="Example: P101"
    )

    query_input = gr.Textbox(
        label="Ask your question",
        placeholder="Example: Explain my lab test report."
    )

    output_box = gr.Textbox(
        label="Assistant Response",
        lines=24
    )

    submit_btn = gr.Button("Get Healthcare Summary")

    submit_btn.click(
        fn=healthcare_support_ui,
        inputs=[patient_input, query_input],
        outputs=output_box
    )

demo.launch(share=True, debug=True)
```
