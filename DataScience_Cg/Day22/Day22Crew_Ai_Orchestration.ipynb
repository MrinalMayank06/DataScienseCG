```python
# Scenario: AI Content Automation for a Startup
# 🏢 Problem Statement

# A startup wants to:

# Stay updated with latest AI trends

# Automatically create LinkedIn posts

# Save time on manual research & writing

# 👉 So we build a multi-agent AI system to automate this using CrewAI.


!pip install crewai litellm langchain-community langchain-litellm
```

```python
import os
os.environ["CREWAI_TRACING_ENABLED"] = "true"

from crewai import Agent, Task, Crew
from google.colab import userdata
from langchain_litellm import ChatLiteLLM
from IPython.display import display, HTML

# -----------------------------
# 1. Set Groq API Key (cleaned)
# -----------------------------
raw_key = userdata.get("GROQ_API_KEY")
clean_key = str(raw_key).strip().replace("\n", "").replace("\r", "")
os.environ["GROQ_API_KEY"] = clean_key

# remove OpenAI key if exists
os.environ.pop("OPENAI_API_KEY", None)

# optional debug
print("API KEY LOADED:", repr(clean_key[:12] + "..."))

# -----------------------------
# 2. Take User Input Prompt
# -----------------------------
user_topic = input("Enter the topic for LinkedIn post / research: ").strip()

if not user_topic:
    raise ValueError("Topic cannot be empty. Please enter a valid topic.")

# -----------------------------
# 3. Define LLM
# -----------------------------
llm = ChatLiteLLM(
    model="groq/llama-3.3-70b-versatile",
    temperature=0.7
)

# -----------------------------
# 4. Define Agents
# -----------------------------
researcher = Agent(
    role="Researcher",
    goal="Find useful and recent insights on the given topic",
    backstory="Expert in research, trends, and summarization",
    llm=llm,
    verbose=True
)

writer = Agent(
    role="Writer",
    goal="Write a strong LinkedIn post from the research",
    backstory="Expert content creator for professional social media posts",
    llm=llm,
    verbose=True
)

# -----------------------------
# 5. Tasks
# -----------------------------
research_task = Task(
    description=f"""
    Research the topic: "{user_topic}"

    Find 5 strong insights, trends, or key points.
    Keep the research clear, practical, and professional.
    """,
    expected_output="A bullet list of 5 researched insights with short summaries.",
    agent=researcher
)

write_task = Task(
    description=f"""
    Using the research, write a polished LinkedIn post on: "{user_topic}"

    Requirements:
    - Start with a strong hook
    - Keep it engaging and professional
    - Add practical insight
    - End with a CTA
    - Make it suitable for LinkedIn audience
    """,
    expected_output="A final LinkedIn post in plain text.",
    agent=writer,
    context=[research_task]
)

# -----------------------------
# 6. Create Crew
# -----------------------------
crew = Crew(
    agents=[researcher, writer],
    tasks=[research_task, write_task],
    verbose=True
)

# -----------------------------
# 7. Run Crew
# -----------------------------
result = crew.kickoff()

final_output = str(result)

print("\n✅ FINAL OUTPUT:\n")
print(final_output)

# -----------------------------
# 8. Convert Output to HTML
# -----------------------------
html_output = f"""
<div style="
    max-width: 900px;
    margin: 20px auto;
    padding: 24px;
    border-radius: 16px;
    background: #f8f9fa;
    border: 1px solid #dcdcdc;
    font-family: Arial, sans-serif;
    line-height: 1.7;
    color: #222;
">
    <h2 style="color:#0A66C2; margin-bottom: 12px;">
        LinkedIn Post Generator Output
    </h2>
    <p style="font-size:16px; margin-bottom: 10px;">
        <strong>Topic:</strong> {user_topic}
    </p>
    <div style="
        white-space: pre-wrap;
        background: white;
        padding: 18px;
        border-radius: 12px;
        border: 1px solid #e0e0e0;
        font-size: 15px;
    ">
        {final_output}
    </div>
</div>
"""

display(HTML(html_output))
```
