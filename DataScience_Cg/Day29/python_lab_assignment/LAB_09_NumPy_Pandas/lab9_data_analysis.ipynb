# LAB 9: NumPy + Pandas

**Objective:** Perform Data Analysis

## Tasks
- Load dataset
- Handle missing values
- Perform filtering
- Compute mean, median, aggregation

```python

import pandas as pd
import numpy as np
from pathlib import Path

data = pd.DataFrame({
    "Student": ["Aman", "Bhavna", "Chetan", "Divya", "Esha", "Farhan"],
    "Marks": [88, np.nan, 79, 92, 85, np.nan],
    "Attendance": [90, 85, np.nan, 95, 88, 80],
    "Department": ["CSE", "ECE", "CSE", "CSE", "ECE", "ME"]
})

csv_path = Path("student_data.csv")
data.to_csv(csv_path, index=False)

df = pd.read_csv(csv_path)
print("Loaded Dataset:")
print(df)
```

```python

print("Missing Values Before Handling:")
print(df.isnull().sum())

df["Marks"] = df["Marks"].fillna(df["Marks"].mean())
df["Attendance"] = df["Attendance"].fillna(df["Attendance"].mean())

print("\nDataset After Handling Missing Values:")
print(df)
```

```python

filtered_df = df[df["Department"] == "CSE"]

print("Filtered Data (Department = CSE):")
print(filtered_df)

print("\nStatistics:")
print("Mean Marks:", df["Marks"].mean())
print("Median Marks:", df["Marks"].median())
print("\nAggregation by Department:")
print(df.groupby("Department")[["Marks", "Attendance"]].mean())
```
