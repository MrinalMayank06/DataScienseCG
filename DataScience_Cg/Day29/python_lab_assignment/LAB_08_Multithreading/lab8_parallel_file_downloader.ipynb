# LAB 8: Multithreading

**Objective:** Simulate Parallel File Downloader

## Tasks
- Download multiple files using threads
- Measure execution time
- Compare with sequential execution

```python

from pathlib import Path
import threading
import time

source_dir = Path("source_files")
download_dir_seq = Path("downloads_sequential")
download_dir_thr = Path("downloads_threaded")
source_dir.mkdir(exist_ok=True)
download_dir_seq.mkdir(exist_ok=True)
download_dir_thr.mkdir(exist_ok=True)

for i in range(1, 4):
    (source_dir / f"file_{i}.txt").write_text(f"Dummy content for file {i}\n" * 5000, encoding="utf-8")

def simulate_download(source, destination):
    content = source.read_text(encoding="utf-8")
    time.sleep(0.2)
    destination.write_text(content, encoding="utf-8")
```

```python

# Sequential execution
start = time.perf_counter()
for file_path in source_dir.glob("*.txt"):
    simulate_download(file_path, download_dir_seq / file_path.name)
sequential_time = time.perf_counter() - start

print(f"Sequential download time: {sequential_time:.4f} seconds")
print("Sequential downloads:", sorted([p.name for p in download_dir_seq.glob('*.txt')]))
```

```python

# Threaded execution
threads = []
start = time.perf_counter()
for file_path in source_dir.glob("*.txt"):
    thread = threading.Thread(
        target=simulate_download,
        args=(file_path, download_dir_thr / file_path.name)
    )
    threads.append(thread)
    thread.start()

for thread in threads:
    thread.join()

threaded_time = time.perf_counter() - start

print(f"Threaded download time: {threaded_time:.4f} seconds")
print("Threaded downloads:", sorted([p.name for p in download_dir_thr.glob('*.txt')]))
print("\nComparison:")
print("Threaded is faster than sequential:", threaded_time < sequential_time)
```
