# LAB 5: Exception Handling

**Objective:** Develop a Robust Calculator

## Tasks
- Handle invalid input
- Handle division by zero
- Use try–except–finally
- Create custom exception

```python

class NegativeNumberError(Exception):
    pass

def robust_calculator(a, b, operation):
    try:
        if operation == "divide":
            if b == 0:
                raise ZeroDivisionError("Division by zero is not allowed.")
            return a / b
        elif operation == "add":
            return a + b
        elif operation == "subtract":
            return a - b
        elif operation == "multiply":
            return a * b
        elif operation == "sqrt_check":
            if a < 0:
                raise NegativeNumberError("Negative number is not allowed for this custom check.")
            return a ** 0.5
        else:
            raise ValueError("Unsupported operation.")
    except ZeroDivisionError as error:
        print("ZeroDivisionError:", error)
    except NegativeNumberError as error:
        print("Custom Exception:", error)
    except ValueError as error:
        print("ValueError:", error)
    finally:
        print("Calculation attempt completed.")
```

```python

print("Add:", robust_calculator(12, 5, "add"))
print("Subtract:", robust_calculator(12, 5, "subtract"))
print("Multiply:", robust_calculator(12, 5, "multiply"))
print("Divide:", robust_calculator(12, 4, "divide"))
print("Divide by Zero Test:", robust_calculator(12, 0, "divide"))
print("Custom Exception Test:", robust_calculator(-9, 0, "sqrt_check"))
print("Invalid Input Test:", robust_calculator(10, 2, "power"))
```
