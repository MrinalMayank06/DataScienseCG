# LAB 6: OOP + Inheritance

**Objective:** Design an Employee Management System

## Tasks
- Base class: Employee
- Derived: Manager, Developer
- Override salary method
- Display employee details

```python

class Employee:
    def __init__(self, emp_id, name, base_salary):
        self.emp_id = emp_id
        self.name = name
        self.base_salary = base_salary

    def calculate_salary(self):
        return self.base_salary

    def display_details(self):
        print(f"ID: {self.emp_id}, Name: {self.name}, Salary: {self.calculate_salary():.2f}")

class Manager(Employee):
    def calculate_salary(self):
        return self.base_salary + 20000

class Developer(Employee):
    def calculate_salary(self):
        return self.base_salary + 10000
```

```python

employees = [
    Manager(101, "Ananya", 70000),
    Developer(102, "Vikram", 60000),
    Developer(103, "Sneha", 65000),
]

print("Employee Details:")
for emp in employees:
    emp.display_details()
```
