# LAB 4: Regex + File Processing

**Objective:** Build an Email Extractor Tool

## Tasks
- Read a text file
- Extract valid emails using regex
- Count unique emails
- Save results in CSV

```python

from pathlib import Path
import re

sample_text = '''
Contact us at support@example.com or sales@example.com.
You can also reach admin@college.edu and hr@company.org.
Duplicate email: support@example.com
Invalid examples: user@@test.com, test@, @domain.com
'''

text_path = Path("emails_source.txt")
text_path.write_text(sample_text, encoding="utf-8")

content = text_path.read_text(encoding="utf-8")
pattern = r'[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Za-z]{2,}'
emails = re.findall(pattern, content)

print("Extracted Emails:")
print(emails)
print("Total Emails Found:", len(emails))
```

```python

import csv

unique_emails = sorted(set(emails))
print("\nUnique Emails:")
print(unique_emails)
print("Unique Email Count:", len(unique_emails))

csv_path = "extracted_emails.csv"
with open(csv_path, "w", newline="", encoding="utf-8") as file:
    writer = csv.writer(file)
    writer.writerow(["Email"])
    for email in unique_emails:
        writer.writerow([email])

print("\nCSV file saved as:", csv_path)
```
