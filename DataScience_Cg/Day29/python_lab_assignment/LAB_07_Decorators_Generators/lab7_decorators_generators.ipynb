# LAB 7: Decorators + Generators

**Objective:** Optimize performance using advanced Python

## Tasks
- Create decorator for logging
- Create generator for large data
- Compare memory usage

```python

import sys
from functools import wraps

def log_function_call(func):
    @wraps(func)
    def wrapper(*args, **kwargs):
        print(f"[LOG] Running function: {func.__name__}")
        result = func(*args, **kwargs)
        print(f"[LOG] Finished function: {func.__name__}")
        return result
    return wrapper

@log_function_call
def sum_of_numbers(n):
    return sum(range(1, n + 1))

def large_data_generator(limit):
    for i in range(limit):
        yield i * i
```

```python

result = sum_of_numbers(10)
print("Sum Result:", result)

generated_data = large_data_generator(100000)
sample_values = [next(generated_data) for _ in range(5)]
print("Generator Sample Output:", sample_values)

list_version = [i * i for i in range(100000)]
generator_version = large_data_generator(100000)

print("\nMemory Usage Comparison:")
print("List Size (bytes):", sys.getsizeof(list_version))
print("Generator Size (bytes):", sys.getsizeof(generator_version))
```
