# LAB 1: Python Fundamentals + File Handling

**Objective:** Build a Student Result Processing System

## Tasks
- Take input for 5 students (name + marks)
- Calculate percentage & grade
- Store results in a .txt file
- Read file and display toppers

```python

from pathlib import Path

students = [
    {"name": "Aarav", "marks": 91},
    {"name": "Diya", "marks": 84},
    {"name": "Kabir", "marks": 76},
    {"name": "Meera", "marks": 95},
    {"name": "Rohan", "marks": 68},
]

def grade_from_marks(marks):
    if marks >= 90:
        return "A+"
    elif marks >= 80:
        return "A"
    elif marks >= 70:
        return "B"
    elif marks >= 60:
        return "C"
    else:
        return "D"

for student in students:
    student["percentage"] = float(student["marks"])
    student["grade"] = grade_from_marks(student["marks"])

students = sorted(students, key=lambda x: x["marks"], reverse=True)
results_path = Path("student_results.txt")

with open(results_path, "w", encoding="utf-8") as file:
    file.write("STUDENT RESULT PROCESSING SYSTEM\n")
    file.write("=" * 40 + "\n")
    for s in students:
        file.write(f"Name: {s['name']}, Marks: {s['marks']}, Percentage: {s['percentage']}%, Grade: {s['grade']}\n")

print("Structured file created:", results_path)
print("\nSorted Results:")
for s in students:
    print(s)
```

```python

from pathlib import Path

results_path = Path("student_results.txt")

print("Reading file content...\n")
print(results_path.read_text(encoding="utf-8"))

print("Topper(s):")
top_marks = max(student["marks"] for student in students)
toppers = [student for student in students if student["marks"] == top_marks]
for topper in toppers:
    print(f"{topper['name']} with {topper['marks']} marks")
```
