# LAB 10: Mini Project (FINAL)

**Objective:** Build a Student Management System (End-to-End)

## Tasks
- Add student
- Search student
- Store in CSV
- Display analytics (topper, avg marks)

```python

import pandas as pd
from pathlib import Path

csv_path = Path("student_management_data.csv")
columns = ["RollNo", "Name", "Marks", "Department"]

def initialize_storage():
    if not csv_path.exists():
        pd.DataFrame(columns=columns).to_csv(csv_path, index=False)

def add_student(roll_no, name, marks, department):
    df = pd.read_csv(csv_path)
    new_row = pd.DataFrame([[roll_no, name, marks, department]], columns=columns)
    df = pd.concat([df, new_row], ignore_index=True)
    df.to_csv(csv_path, index=False)

def search_student(name):
    df = pd.read_csv(csv_path)
    return df[df["Name"].str.lower() == name.lower()]

def show_analytics():
    df = pd.read_csv(csv_path)
    topper = df.loc[df["Marks"].idxmax()]
    average_marks = df["Marks"].mean()
    return topper, average_marks, df

initialize_storage()
add_student(1, "Aditi", 89, "CSE")
add_student(2, "Karan", 76, "ECE")
add_student(3, "Neha", 94, "CSE")
add_student(4, "Yash", 81, "ME")

print("Student records stored in CSV.")
print(pd.read_csv(csv_path))
```

```python

print("Search Result for 'Neha':")
print(search_student("Neha"))
```

```python

topper, average_marks, df = show_analytics()
print("Topper:")
print(topper)
print("\nAverage Marks:", round(average_marks, 2))
print("\nAll Student Data:")
print(df)
```
