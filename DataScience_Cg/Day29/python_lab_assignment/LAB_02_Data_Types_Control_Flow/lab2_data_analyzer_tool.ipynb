# LAB 2: Data Types + Control Flow

**Objective:** Build a Data Analyzer Tool

## Tasks
- Input a mixed list
- Separate integers, floats, strings
- Count frequency of each type
- Print summary using loops

```python

mixed_list = [10, 3.14, "apple", 7, "banana", 2.71, 15, "apple", 0.5, "data"]

integers = []
floats = []
strings = []

for item in mixed_list:
    if isinstance(item, int):
        integers.append(item)
    elif isinstance(item, float):
        floats.append(item)
    elif isinstance(item, str):
        strings.append(item)

print("Original Mixed List:")
print(mixed_list)
print("\nSeparated Values:")
print("Integers:", integers)
print("Floats:", floats)
print("Strings:", strings)
```

```python

type_frequency = {
    "int": len(integers),
    "float": len(floats),
    "str": len(strings),
}

print("Type Frequency Summary:")
for data_type, count in type_frequency.items():
    print(f"{data_type}: {count}")

print("\nDetailed Loop Summary:")
for i, value in enumerate(mixed_list, start=1):
    print(f"Index {i}: {value} -> {type(value).__name__}")
```
