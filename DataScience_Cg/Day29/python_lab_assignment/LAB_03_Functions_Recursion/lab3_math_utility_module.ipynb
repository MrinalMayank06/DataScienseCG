# LAB 3: Functions + Recursion

**Objective:** Create a Math Utility Module

## Tasks
- Factorial using recursion
- Fibonacci using recursion
- Lambda for square & cube
- Compare execution time

```python

import time

def factorial(n):
    if n == 0 or n == 1:
        return 1
    return n * factorial(n - 1)

def fibonacci(n):
    if n <= 1:
        return n
    return fibonacci(n - 1) + fibonacci(n - 2)

square = lambda x: x ** 2
cube = lambda x: x ** 3

number = 5
print(f"Factorial of {number} =", factorial(number))
print(f"Fibonacci of {number} =", fibonacci(number))
print(f"Square of {number} =", square(number))
print(f"Cube of {number} =", cube(number))
```

```python

factorial_input = 10
fibonacci_input = 10

start = time.perf_counter()
factorial_result = factorial(factorial_input)
factorial_time = time.perf_counter() - start

start = time.perf_counter()
fibonacci_result = fibonacci(fibonacci_input)
fibonacci_time = time.perf_counter() - start

print("Execution Time Comparison:")
print(f"Factorial({factorial_input}) = {factorial_result} | Time: {factorial_time:.8f} seconds")
print(f"Fibonacci({fibonacci_input}) = {fibonacci_result} | Time: {fibonacci_time:.8f} seconds")
```
